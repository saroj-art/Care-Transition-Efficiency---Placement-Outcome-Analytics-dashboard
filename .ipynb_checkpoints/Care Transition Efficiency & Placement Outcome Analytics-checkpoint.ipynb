## Phase 1 — Initial Data Quality Assessment

```python
# ============================================================
# Step 10: Missing value analysis
# ============================================================

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

missing_summary = missing_summary.sort_values(
    by="Missing_Count",
    ascending=False
)

missing_summary
```

```python
# ============================================================
# Step 11: Duplicate record analysis
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# ============================================================
# Step 12: Duplicate Date analysis
# ============================================================

duplicate_dates = df["Date"].duplicated().sum()

print("Number of duplicate Date values:", duplicate_dates)
```

```python
# ============================================================
# Step 14: Check for negative numerical values
# ============================================================

numeric_columns = df.select_dtypes(include=np.number).columns

negative_summary = {}

for column in numeric_columns:
    negative_summary[column] = (df[column] < 0).sum()

negative_summary = pd.Series(negative_summary, name="Negative_Values")

negative_summary
```

```python
# ============================================================
# Step 15: Check for zero values
# ============================================================

zero_summary = {}

for column in numeric_columns:
    zero_summary[column] = (df[column] == 0).sum()

zero_summary = pd.Series(zero_summary, name="Zero_Values")

zero_summary
```

```python
# ============================================================
# PHASE 2: DATA CLEANING
# Step 1: Remove completely empty rows
# ============================================================

df_cleaned = df.dropna(how="all").copy()

print("Original shape:", df.shape)
print("Shape after removing completely empty rows:", df_cleaned.shape)

print("\nRemaining missing values:")
print(df_cleaned.isnull().sum())
```

Original shape: (1170, 6)
Shape after removing completely empty rows: (720, 6)

```python
# ============================================================
# Step 2: Standardize column names
# ============================================================

df_cleaned.columns = [
    "Date",
    "CBP_Intake",
    "CBP_Custody",
    "CBP_Transfers",
    "HHS_Care",
    "HHS_Discharges"
]

print(df_cleaned.columns.tolist())
```

```python
# ============================================================
# Step 3: Convert Date to datetime
# ============================================================

df_cleaned["Date"] = pd.to_datetime(
    df_cleaned["Date"],
    errors="coerce"
)

print("Invalid dates:", df_cleaned["Date"].isna().sum())
print("Minimum Date:", df_cleaned["Date"].min())
print("Maximum Date:", df_cleaned["Date"].max())
```

```python
# ============================================================
# Step 4: Convert numerical columns
# ============================================================

numeric_columns = [
    "CBP_Intake",
    "CBP_Custody",
    "CBP_Transfers",
    "HHS_Care",
    "HHS_Discharges"
]

for column in numeric_columns:
    df_cleaned[column] = (
        df_cleaned[column]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    df_cleaned[column] = pd.to_numeric(
        df_cleaned[column],
        errors="coerce"
    )

print(df_cleaned.dtypes)
```

```python
# ============================================================
# Step 5: Final missing-value check
# ============================================================

missing_summary = pd.DataFrame({
    "Missing_Count": df_cleaned.isnull().sum(),
    "Missing_Percentage": (
        df_cleaned.isnull().sum() / len(df_cleaned)
    ) * 100
})

missing_summary
```

```python
# ============================================================
# Step 6: Duplicate check
# ============================================================

print(
    "Duplicate rows:",
    df_cleaned.duplicated().sum()
)

print(
    "Duplicate dates:",
    df_cleaned["Date"].duplicated().sum()
)
```

```python
# ============================================================
# Step 7: Negative-value validation
# ============================================================

negative_summary = (df_cleaned[numeric_columns] < 0).sum()

negative_summary
```

```python
# ============================================================
# Step 8: Zero-value check
# ============================================================

zero_summary = (df_cleaned[numeric_columns] == 0).sum()

zero_summary
# We will keep these zeros for now.
```

```python
# ============================================================
# Step 9: Sort chronologically
# ============================================================

df_cleaned = (
    df_cleaned
    .sort_values("Date")
    .reset_index(drop=True)
)

df_cleaned.head()
```

## PHASE 5 — PIPELINE FLOW METRICS

```python
# ============================================================
# PHASE 5.1: Pipeline entries, exits and net flow
# ============================================================

df_pipeline["Pipeline_Entries"] = (
    df_pipeline["CBP_Intake"]
)

df_pipeline["Pipeline_Exits"] = (
    df_pipeline["HHS_Discharges"]
)

df_pipeline["Net_Pipeline_Flow"] = (
    df_pipeline["Pipeline_Entries"]
    - df_pipeline["Pipeline_Exits"]
)

df_pipeline["Positive_Accumulation"] = (
    df_pipeline["Net_Pipeline_Flow"]
    .clip(lower=0)
)

df_pipeline["Negative_Accumulation"] = (
    df_pipeline["Net_Pipeline_Flow"]
    .clip(upper=0)
)
```

## PHASE 6 — REQUIRED KPIs

```python
# ============================================================
# PHASE 6.1: Transfer Efficiency Ratio
# Transfers / CBP Custody
# ============================================================

df_pipeline["Transfer_Efficiency_Ratio"] = np.where(
    df_pipeline["CBP_Custody"] > 0,
    df_pipeline["CBP_Transfers"]
    / df_pipeline["CBP_Custody"],
    np.nan
)
```

```python
# ============================================================
# PHASE 6.2: Discharge Effectiveness Index
# Discharges / HHS Care
# ============================================================

df_pipeline["Discharge_Effectiveness_Index"] = np.where(
    df_pipeline["HHS_Care"] > 0,
    df_pipeline["HHS_Discharges"]
    / df_pipeline["HHS_Care"],
    np.nan
)
```

```python
# ============================================================
# PHASE 6.3: Pipeline Throughput
# Exits / Entries
# ============================================================

df_pipeline["Daily_Pipeline_Throughput"] = np.where(
    df_pipeline["Pipeline_Entries"] > 0,
    df_pipeline["Pipeline_Exits"]
    / df_pipeline["Pipeline_Entries"],
    np.nan
)
```

```python
# ============================================================
# PHASE 6.4: KPI percentages
# ============================================================

df_pipeline["Transfer_Efficiency_Pct"] = (
    df_pipeline["Transfer_Efficiency_Ratio"] * 100
)

df_pipeline["Discharge_Effectiveness_Pct"] = (
    df_pipeline["Discharge_Effectiveness_Index"] * 100
)

df_pipeline["Pipeline_Throughput_Pct"] = (
    df_pipeline["Daily_Pipeline_Throughput"] * 100
)

print("All core KPIs created successfully.")
```

```python
# ============================================================
# Cumulative backlog-pressure proxy
# ============================================================

df_pipeline["Cumulative_Backlog_Pressure"] = (
    df_pipeline["Net_Pipeline_Flow"].cumsum()
)

df_pipeline["Backlog_Accumulation_Rate"] = np.where(
    df_pipeline["Pipeline_Entries"] > 0,
    df_pipeline["Positive_Accumulation"]
    / df_pipeline["Pipeline_Entries"],
    0
)

df_pipeline["Backlog_Accumulation_Rate_Pct"] = (
    df_pipeline["Backlog_Accumulation_Rate"] * 100
)
```

## PHASE 13 — MONTHLY ANALYSIS

```python
# ============================================================
# PHASE 13: MONTHLY AGGREGATION
# ============================================================

monthly_df = (
    df_pipeline
    .groupby("Month_Period")
    .agg(
        CBP_Intake=("CBP_Intake", "sum"),
        CBP_Transfers=("CBP_Transfers", "sum"),
        HHS_Discharges=("HHS_Discharges", "sum"),

        CBP_Custody_Total=("CBP_Custody", "sum"),
        HHS_Care_Total=("HHS_Care", "sum"),

        CBP_Custody_Avg=("CBP_Custody", "mean"),
        HHS_Care_Avg=("HHS_Care", "mean"),

        Reporting_Days=("Date", "count")
    )
    .reset_index()
)

monthly_df["Transfer_Efficiency"] = np.where(
    monthly_df["CBP_Custody_Total"] > 0,
    monthly_df["CBP_Transfers"]
    / monthly_df["CBP_Custody_Total"],
    np.nan
)

monthly_df["Discharge_Effectiveness"] = np.where(
    monthly_df["HHS_Care_Total"] > 0,
    monthly_df["HHS_Discharges"]
    / monthly_df["HHS_Care_Total"],
    np.nan
)

monthly_df["Pipeline_Throughput"] = np.where(
    monthly_df["CBP_Intake"] > 0,
    monthly_df["HHS_Discharges"]
    / monthly_df["CBP_Intake"],
    np.nan
)

monthly_df["Net_Pipeline_Flow"] = (
    monthly_df["CBP_Intake"]
    - monthly_df["HHS_Discharges"]
)

monthly_df["Backlog_Accumulation_Rate"] = np.where(
    monthly_df["CBP_Intake"] > 0,
    monthly_df["Net_Pipeline_Flow"].clip(lower=0)
    / monthly_df["CBP_Intake"],
    0
)

monthly_df
```

## PHASE 17 — DAY-OF-WEEK PATTERN

```python
# ============================================================
# Day-of-week analysis
# ============================================================

dayofweek_summary = (
    df_pipeline
    .groupby(["Day_of_Week", "Weekday"])
    .agg(
        Avg_Intake=("CBP_Intake", "mean"),
        Avg_Transfers=("CBP_Transfers", "mean"),
        Avg_Discharges=("HHS_Discharges", "mean"),
        Avg_Transfer_Efficiency=(
            "Transfer_Efficiency_Ratio",
            "mean"
        ),
        Avg_Discharge_Effectiveness=(
            "Discharge_Effectiveness_Index",
            "mean"
        )
    )
    .reset_index()
    .sort_values("Day_of_Week")
)

dayofweek_summary
```

## PHASE 18 — ROLLING PERFORMANCE

```python
# ============================================================
# PHASE 18: ROLLING PERFORMANCE
# ============================================================

df_pipeline["Transfer_Efficiency_7Obs_Rolling"] = (
    df_pipeline["Transfer_Efficiency_Ratio"]
    .rolling(
        window=7,
        min_periods=3
    )
    .mean()
)

df_pipeline["Discharge_Effectiveness_7Obs_Rolling"] = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .rolling(
        window=7,
        min_periods=3
    )
    .mean()
)

df_pipeline["Throughput_7Obs_Rolling"] = (
    df_pipeline["Daily_Pipeline_Throughput"]
    .rolling(
        window=7,
        min_periods=3
    )
    .mean()
)
```

## PHASE 19 — BOTTLENECK DETECTION

```python
# ============================================================
# PHASE 19: BOTTLENECK THRESHOLDS
# ============================================================

transfer_low_threshold = (
    df_pipeline["Transfer_Efficiency_Ratio"]
    .quantile(0.25)
)

discharge_low_threshold = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .quantile(0.25)
)

high_cbp_custody_threshold = (
    df_pipeline["CBP_Custody"]
    .quantile(0.75)
)

high_hhs_care_threshold = (
    df_pipeline["HHS_Care"]
    .quantile(0.75)
)

print("Low transfer threshold:",
      transfer_low_threshold)

print("Low discharge threshold:",
      discharge_low_threshold)

print("High CBP custody threshold:",
      high_cbp_custody_threshold)

print("High HHS care threshold:",
      high_hhs_care_threshold)
```

```python
# CBP transfer bottleneck
df_pipeline["CBP_Transfer_Bottleneck_Flag"] = (
    (df_pipeline["Transfer_Efficiency_Ratio"]
     <= transfer_low_threshold)
    &
    (df_pipeline["CBP_Custody"]
     >= high_cbp_custody_threshold)
)
```

```python
# HHS discharge bottleneck
df_pipeline["HHS_Discharge_Bottleneck_Flag"] = (
    (df_pipeline["Discharge_Effectiveness_Index"]
     <= discharge_low_threshold)
    &
    (df_pipeline["HHS_Care"]
     >= high_hhs_care_threshold)
)
```

```python
# Overall bottleneck
df_pipeline["Any_Bottleneck_Flag"] = (
    df_pipeline["CBP_Transfer_Bottleneck_Flag"]
    |
    df_pipeline["HHS_Discharge_Bottleneck_Flag"]
)

print(
    "CBP transfer bottleneck observations:",
    df_pipeline["CBP_Transfer_Bottleneck_Flag"].sum()
)

print(
    "HHS discharge bottleneck observations:",
    df_pipeline["HHS_Discharge_Bottleneck_Flag"].sum()
)

print(
    "Total bottleneck observations:",
    df_pipeline["Any_Bottleneck_Flag"].sum()
)
```

## PHASE 20 — BOTTLENECK OUTPUT

```python
bottleneck_periods = (
    df_pipeline[
        df_pipeline["Any_Bottleneck_Flag"]
    ]
    [
        [
            "Date",
            "CBP_Custody",
            "CBP_Transfers",
            "Transfer_Efficiency_Pct",
            "HHS_Care",
            "HHS_Discharges",
            "Discharge_Effectiveness_Pct",
            "Net_Pipeline_Flow",
            "Cumulative_Backlog_Pressure"
        ]
    ]
    .sort_values("Date")
)

display(bottleneck_periods.head(30))
```

## PHASE 21 — PROLONGED STAGNATION


```python
# ============================================================
# PHASE 21: PROLONGED STAGNATION
# ============================================================

df_pipeline["Transfer_Stagnation_Flag"] = (
    df_pipeline["Transfer_Efficiency_Ratio"]
    <= transfer_low_threshold
)

df_pipeline["Discharge_Stagnation_Flag"] = (
    df_pipeline["Discharge_Effectiveness_Index"]
    <= discharge_low_threshold
)
```

```python
# Consecutive groups
transfer_group = (
    df_pipeline["Transfer_Stagnation_Flag"]
    .ne(
        df_pipeline["Transfer_Stagnation_Flag"].shift()
    )
    .cumsum()
)

discharge_group = (
    df_pipeline["Discharge_Stagnation_Flag"]
    .ne(
        df_pipeline["Discharge_Stagnation_Flag"].shift()
    )
    .cumsum()
)

df_pipeline["Transfer_Stagnation_Group"] = transfer_group
df_pipeline["Discharge_Stagnation_Group"] = discharge_group
```

```python
# Transfer stagnation periods
transfer_stagnation_summary = (
    df_pipeline[
        df_pipeline["Transfer_Stagnation_Flag"]
    ]
    .groupby("Transfer_Stagnation_Group")
    .agg(
        Start_Date=("Date", "min"),
        End_Date=("Date", "max"),
        Observations=("Date", "count"),
        Avg_Transfer_Efficiency=(
            "Transfer_Efficiency_Ratio",
            "mean"
        ),
        Avg_CBP_Custody=("CBP_Custody", "mean")
    )
    .reset_index()
)

transfer_stagnation_summary = (
    transfer_stagnation_summary[
        transfer_stagnation_summary["Observations"] >= 3
    ]
    .sort_values(
        "Observations",
        ascending=False
    )
)

transfer_stagnation_summary
```

```python
# Discharge stagnation periods
discharge_stagnation_summary = (
    df_pipeline[
        df_pipeline["Discharge_Stagnation_Flag"]
    ]
    .groupby("Discharge_Stagnation_Group")
    .agg(
        Start_Date=("Date", "min"),
        End_Date=("Date", "max"),
        Observations=("Date", "count"),
        Avg_Discharge_Effectiveness=(
            "Discharge_Effectiveness_Index",
            "mean"
        ),
        Avg_HHS_Care=("HHS_Care", "mean")
    )
    .reset_index()
)

discharge_stagnation_summary = (
    discharge_stagnation_summary[
        discharge_stagnation_summary["Observations"] >= 3
    ]
    .sort_values(
        "Observations",
        ascending=False
    )
)

discharge_stagnation_summary
```

## PHASE 22 — OUTCOME STABILITY

```python
# ============================================================
# PHASE 22: OUTCOME STABILITY
# ============================================================

discharge_mean = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .mean()
)

discharge_std = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .std()
)

discharge_cv = (
    discharge_std / discharge_mean
    if discharge_mean != 0
    else np.nan
)

print("Mean:", discharge_mean)
print("Standard deviation:", discharge_std)
print("Coefficient of variation:", discharge_cv)
```

```python
# Stability score 
# ============================================================
# Outcome Stability Score
# Higher score = greater consistency
# ============================================================

rolling_mean = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .rolling(
        window=7,
        min_periods=3
    )
    .mean()
)

rolling_std = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .rolling(
        window=7,
        min_periods=3
    )
    .std()
)

rolling_cv = np.where(
    rolling_mean > 0,
    rolling_std / rolling_mean,
    np.nan
)

df_pipeline["Outcome_Stability_Score"] = (
    100 / (1 + rolling_cv)
)

df_pipeline["Outcome_Stability_Score"] = (
    df_pipeline["Outcome_Stability_Score"]
    .clip(0, 100)
)
```

```python
# Stability chart
plt.figure(figsize=(14, 5))

plt.plot(
    df_pipeline["Date"],
    df_pipeline["Outcome_Stability_Score"]
)

plt.title("Outcome Stability Score")
plt.xlabel("Date")
plt.ylabel("Stability Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## PHASE 23 — VARIABILITY ANALYSIS

```python
# ============================================================
# PHASE 23: VARIABILITY
# ============================================================

variability_summary = pd.DataFrame({
    "Metric": [
        "CBP Intake",
        "CBP Transfers",
        "HHS Discharges",
        "Transfer Efficiency",
        "Discharge Effectiveness",
        "Pipeline Throughput"
    ],

    "Mean": [
        df_pipeline["CBP_Intake"].mean(),
        df_pipeline["CBP_Transfers"].mean(),
        df_pipeline["HHS_Discharges"].mean(),
        df_pipeline["Transfer_Efficiency_Ratio"].mean(),
        df_pipeline["Discharge_Effectiveness_Index"].mean(),
        df_pipeline["Daily_Pipeline_Throughput"].mean()
    ],

    "Std_Dev": [
        df_pipeline["CBP_Intake"].std(),
        df_pipeline["CBP_Transfers"].std(),
        df_pipeline["HHS_Discharges"].std(),
        df_pipeline["Transfer_Efficiency_Ratio"].std(),
        df_pipeline["Discharge_Effectiveness_Index"].std(),
        df_pipeline["Daily_Pipeline_Throughput"].std()
    ]
})

variability_summary["CV"] = (
    variability_summary["Std_Dev"]
    / variability_summary["Mean"]
)

variability_summary
```

## PHASE 26 — OPERATIONAL ANOMALIES

```python
# ============================================================
# PHASE 26: Z-SCORE ANOMALY DETECTION
# ============================================================

for column in core_columns:

    mean_value = df_pipeline[column].mean()
    std_value = df_pipeline[column].std()

    if std_value > 0:

        df_pipeline[f"{column}_ZScore"] = (
            df_pipeline[column] - mean_value
        ) / std_value

    else:

        df_pipeline[f"{column}_ZScore"] = 0
```

```python
zscore_columns = [
    f"{column}_ZScore"
    for column in core_columns
]

df_pipeline["Operational_Anomaly_Flag"] = (
    df_pipeline[zscore_columns]
    .abs()
    .gt(3)
    .any(axis=1)
)

print(
    "Operational anomaly observations:",
    df_pipeline["Operational_Anomaly_Flag"].sum()
)
```

```python
anomaly_rows = df_pipeline[
    df_pipeline["Operational_Anomaly_Flag"]
][
    [
        "Date",
        "CBP_Intake",
        "CBP_Custody",
        "CBP_Transfers",
        "HHS_Care",
        "HHS_Discharges"
    ]
]

display(anomaly_rows)
```

## PHASE 27 — HIGH PRESSURE PERIODS

```python
# ============================================================
# PHASE 27: HIGH PRESSURE CONDITIONS
# ============================================================

high_intake_threshold = (
    df_pipeline["CBP_Intake"]
    .quantile(0.75)
)

high_hhs_care_threshold = (
    df_pipeline["HHS_Care"]
    .quantile(0.75)
)

low_transfer_threshold = (
    df_pipeline["Transfer_Efficiency_Ratio"]
    .quantile(0.25)
)

low_discharge_threshold = (
    df_pipeline["Discharge_Effectiveness_Index"]
    .quantile(0.25)
)

df_pipeline["High_CBP_Pressure_Flag"] = (
    (df_pipeline["CBP_Intake"]
     >= high_intake_threshold)
    &
    (df_pipeline["Transfer_Efficiency_Ratio"]
     <= low_transfer_threshold)
)

df_pipeline["High_HHS_Pressure_Flag"] = (
    (df_pipeline["HHS_Care"]
     >= high_hhs_care_threshold)
    &
    (df_pipeline["Discharge_Effectiveness_Index"]
     <= low_discharge_threshold)
)

df_pipeline["High_System_Pressure_Flag"] = (
    df_pipeline["High_CBP_Pressure_Flag"]
    |
    df_pipeline["High_HHS_Pressure_Flag"]
)

print(
    "High CBP pressure:",
    df_pipeline["High_CBP_Pressure_Flag"].sum()
)

print(
    "High HHS pressure:",
    df_pipeline["High_HHS_Pressure_Flag"].sum()
)

print(
    "High system pressure:",
    df_pipeline["High_System_Pressure_Flag"].sum()
)
```

## PHASE 28 — KPI SUMMARY

```python
# ============================================================
# PHASE 28: OVERALL KPI SUMMARY
# ============================================================

total_entries = (
    df_pipeline["Pipeline_Entries"].sum()
)

total_exits = (
    df_pipeline["Pipeline_Exits"].sum()
)

total_transfers = (
    df_pipeline["CBP_Transfers"].sum()
)

total_custody_observations = (
    df_pipeline["CBP_Custody"].sum()
)

total_hhs_care_observations = (
    df_pipeline["HHS_Care"].sum()
)

overall_transfer_efficiency = np.where(
    total_custody_observations > 0,
    total_transfers / total_custody_observations,
    np.nan
)

overall_discharge_effectiveness = np.where(
    total_hhs_care_observations > 0,
    total_exits / total_hhs_care_observations,
    np.nan
)

overall_throughput = np.where(
    total_entries > 0,
    total_exits / total_entries,
    np.nan
)

overall_net_flow = (
    total_entries - total_exits
)

overall_backlog_rate = np.where(
    total_entries > 0,
    max(overall_net_flow, 0) / total_entries,
    np.nan
)

overall_stability_score = (
    100 / (1 + discharge_cv)
    if pd.notna(discharge_cv)
    else np.nan
)

kpi_summary = pd.DataFrame({
    "KPI": [
        "Transfer Efficiency Ratio",
        "Discharge Effectiveness Index",
        "Pipeline Throughput",
        "Backlog Accumulation Rate",
        "Outcome Stability Score"
    ],

    "Value": [
        float(overall_transfer_efficiency) * 100,
        float(overall_discharge_effectiveness) * 100,
        float(overall_throughput) * 100,
        float(overall_backlog_rate) * 100,
        float(overall_stability_score)
    ],

    "Unit": [
        "%",
        "%",
        "%",
        "%",
        "Score / 100"
    ]
})

kpi_summary
```

## PHASE 30 — YEARLY ANALYSIS

```python
# ============================================================
# YEARLY ANALYSIS
# ============================================================

yearly_df = (
    df_pipeline
    .groupby("Year")
    .agg(
        Total_Intake=("CBP_Intake", "sum"),
        Total_Transfers=("CBP_Transfers", "sum"),
        Total_Discharges=("HHS_Discharges", "sum"),
        Avg_CBP_Custody=("CBP_Custody", "mean"),
        Avg_HHS_Care=("HHS_Care", "mean"),
        Reporting_Observations=("Date", "count")
    )
    .reset_index()
)

yearly_df["Pipeline_Throughput"] = np.where(
    yearly_df["Total_Intake"] > 0,
    yearly_df["Total_Discharges"]
    / yearly_df["Total_Intake"],
    np.nan
)

yearly_df["Net_Flow"] = (
    yearly_df["Total_Intake"]
    - yearly_df["Total_Discharges"]
)

yearly_df["Pipeline_Throughput_Pct"] = (
    yearly_df["Pipeline_Throughput"] * 100
)

yearly_df
```

## PHASE 31 — QUARTERLY ANALYSIS

```python
quarterly_df = (
    df_pipeline
    .groupby(["Year", "Quarter"])
    .agg(
        Total_Intake=("CBP_Intake", "sum"),
        Total_Transfers=("CBP_Transfers", "sum"),
        Total_Discharges=("HHS_Discharges", "sum"),
        Avg_CBP_Custody=("CBP_Custody", "mean"),
        Avg_HHS_Care=("HHS_Care", "mean")
    )
    .reset_index()
)

quarterly_df["Pipeline_Throughput"] = np.where(
    quarterly_df["Total_Intake"] > 0,
    quarterly_df["Total_Discharges"]
    / quarterly_df["Total_Intake"],
    np.nan
)

quarterly_df["Net_Flow"] = (
    quarterly_df["Total_Intake"]
    - quarterly_df["Total_Discharges"]
)

quarterly_df
```
